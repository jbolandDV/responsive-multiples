```python
import numpy as np
import pandas as pd

# --- helpers
def nonempty(s: pd.Series) -> pd.Series:
    # treats NaN, empty strings, and whitespace as "no touch"
    return s.fillna("").astype(str).str.strip().ne("")

df = df.copy()

# define the touches the way you described
df["touch_mail"] = nonempty(df["appeal_name"])  # mailed
df["touch_email"] = df["campaign_name"].fillna("").astype(str).str.contains("email", case=False, na=False)
df["touch_prog"] = nonempty(df["Last_Impression_Campaign_Name"])  # programmatic last impression exists

# touch count + combo label
df["touch_count"] = df[["touch_mail","touch_email","touch_prog"]].sum(axis=1)

def combo_label(r):
    parts = []
    if r["touch_mail"]:  parts.append("Mail")
    if r["touch_email"]: parts.append("Email")
    if r["touch_prog"]:  parts.append("Programmatic")
    return " + ".join(parts) if parts else "Digital"

df["touch_combo"] = df.apply(combo_label, axis=1)
```

```python
combo_summary = (
    df.groupby("touch_combo", dropna=False)
      .agg(
          transactions=("transaction_id","nunique"),
          donors=("donor_id","nunique"),
          revenue=("amount","sum"),
          avg_gift=("amount","mean"),
          median_gift=("amount","median"),
      )
      .sort_values("transactions", ascending=False)
)

combo_summary["pct_transactions"] = combo_summary["transactions"] / combo_summary["transactions"].sum()
combo_summary["pct_revenue"] = combo_summary["revenue"] / combo_summary["revenue"].sum()

combo_summary
```

```python
import matplotlib.pyplot as plt

plot_df = combo_summary.reset_index()

plt.figure(figsize=(10, 5))
plt.barh(plot_df["touch_combo"], plot_df["transactions"])
plt.xlabel("Transactions")
plt.ylabel("")
plt.title("Multi-touch intersections (Mail / Email / Programmatic)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define bins for 7-day intervals
max_days = df['days_since_in_home'].max()
bins = np.arange(0, max_days + 7, 7)

plt.figure(figsize=(12, 6))
plt.hist(df['days_since_in_home'], bins=bins, edgecolor='black')
plt.title('Distribution of Days Since In-Home (7-day Bins)')
plt.xlabel('Days Since In-Home (Weeks)')
plt.ylabel('Number of Responses')

# Set x-axis ticks and labels to represent weeks
week_labels = [f'{int(i/7)} week(s)' for i in bins if i % 7 == 0]
plt.xticks(bins, week_labels, rotation=45, ha='right')

plt.grid(axis='y', alpha=0.75)
plt.tight_layout()
plt.show()
```

```python
campaign_type_summary = (
    df.groupby(["campaign_type", "touch_combo"], dropna=False)
      .agg(
          transactions=("transaction_id","nunique"),
          donors=("donor_id","nunique"),
          revenue=("amount","sum"),
          avg_gift=("amount","mean"),
          median_gift=("amount","median"),
      )
      .sort_values(["campaign_type", "transactions"], ascending=[True, False])
)

campaign_type_summary["pct_transactions"] = campaign_type_summary.groupby(level=0)["transactions"].transform(lambda x: x / x.sum())
campaign_type_summary["pct_revenue"] = campaign_type_summary.groupby(level=0)["revenue"].transform(lambda x: x / x.sum())

campaign_type_summary
```

```python
appeal_name_summary = (
    df.groupby(["appeal_name", "mail_drop_date", "touch_combo"], dropna=False)
      .agg(
          transactions=("transaction_id","nunique"),
          donors=("donor_id","nunique"),
          revenue=("amount","sum"),
          avg_gift=("amount","mean"),
          median_gift=("amount","median"),
      )
)

# Calculate percentages based on the current grouping
appeal_name_summary["pct_transactions"] = appeal_name_summary.groupby(level=[0,1])["transactions"].transform(lambda x: x / x.sum())
appeal_name_summary["pct_revenue"] = appeal_name_summary.groupby(level=[0,1])["revenue"].transform(lambda x: x / x.sum())

# Calculate totals for each appeal_name and mail_drop_date
appeal_totals = appeal_name_summary.groupby(level=[0,1]).agg(
    transactions=('transactions', 'sum'),
    donors=('donors', 'sum'),
    revenue=('revenue', 'sum'),
    avg_gift=('avg_gift', 'mean'), # Take mean of avg_gift for total, or recalculate if needed
    median_gift=('median_gift', 'mean') # Take mean of median_gift for total, or recalculate if needed
)
appeal_totals['touch_combo'] = 'Total'
appeal_totals['pct_transactions'] = 1.0
appeal_totals['pct_revenue'] = 1.0

# Reset index to allow setting 'touch_combo' as a column, then set it back as part of the index for concatenation
appeal_totals = appeal_totals.reset_index().set_index(['appeal_name', 'mail_drop_date', 'touch_combo'])

# Concatenate the summary with the totals
combined_summary = pd.concat([appeal_name_summary, appeal_totals])

# Reset index to make 'appeal_name', 'mail_drop_date', 'touch_combo' regular columns
combined_summary = combined_summary.reset_index()

# Reorder columns to put 'mail_drop_date' first, then 'appeal_name', 'touch_combo'
cols = combined_summary.columns.tolist()
# Remove the columns to be reordered
cols.remove('mail_drop_date')
cols.remove('appeal_name')
cols.remove('touch_combo')

# Insert them at the desired positions
new_cols = ['mail_drop_date', 'appeal_name', 'touch_combo'] + cols
combined_summary = combined_summary[new_cols]

# Sort by mail_drop_date descending, and then by transactions descending
combined_summary = combined_summary.sort_values(by=['mail_drop_date', 'transactions'], ascending=[False, False])

combined_summary
```

```python
combined_summary.to_csv('combined_summary.csv')
print('combined_summary.csv has been saved.')
```

```python
# Define the bins and labels for Impression_Count
bins = [-1, 8, 20, 50, 100, float('inf')]  # -1 to catch values from 0 up to 4, 5 for 5-10, 10 for 10-20, etc.
labels = ['Less than 7', '8-20', '21-49', '50-100', '100+']

# Create a new column with the impression count buckets
df['impression_count_bucket'] = pd.cut(df['Impression_Count'], bins=bins, labels=labels, right=False)

# Count the number of transactions in each bucket
impression_bucket_summary = df['impression_count_bucket'].value_counts().sort_index()

print("Impression Count Buckets:\n", impression_bucket_summary)

# Optional: Plot the results for better visualization
plt.figure(figsize=(10, 6))
impression_bucket_summary.plot(kind='bar', color='skyblue')
plt.title('Distribution of Impression Counts')
plt.xlabel('Impression Count Bucket')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df_weekly_amount = df.copy()

# Ensure 'transaction_date' is in datetime format
df_weekly_amount['transaction_date'] = pd.to_datetime(df_weekly_amount['transaction_date'])

# Set 'transaction_date' as index for resampling
df_weekly_amount = df_weekly_amount.set_index('transaction_date')

# Resample 'amount' by week and sum the values
weekly_amount = df_weekly_amount['amount'].resample('W').sum()

# Plotting the weekly amount
plt.figure(figsize=(14, 7))
weekly_amount.plot(kind='line', marker='o', linestyle='-')
plt.title('Total Amount by Week')
plt.xlabel('Week')
plt.ylabel('Total Amount')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Prepare df for plotting (current year)
df_weekly_amount = df.copy()
df_weekly_amount['transaction_date'] = pd.to_datetime(df_weekly_amount['transaction_date'])
df_weekly_amount = df_weekly_amount.set_index('transaction_date')
weekly_amount = df_weekly_amount['amount'].resample('W').sum()

# Prepare df_25 for plotting (previous year, aligned to current year)
df_weekly_amount_25 = df_25.copy()
df_weekly_amount_25['transaction_date'] = pd.to_datetime(df_weekly_amount_25['transaction_date'])

# Get the earliest year from the main dataframe (df) to align df_25
# Assuming df represents the current year's data or the year we want to compare against
current_year = df_weekly_amount.index.year.min()

# Adjust the year of df_25 transactions to match the 'current_year' for YoY comparison
def adjust_year(date):
    return date.replace(year=current_year)

df_weekly_amount_25['adjusted_date'] = df_weekly_amount_25['transaction_date'].apply(adjust_year)
df_weekly_amount_25 = df_weekly_amount_25.set_index('adjusted_date')
weekly_amount_25 = df_weekly_amount_25['amount'].resample('W').sum()

# Plotting both weekly amounts on the same graph
plt.figure(figsize=(14, 7))
weekly_amount.plot(kind='line', marker='o', linestyle='-', label='Current Year (df)')
weekly_amount_25.plot(kind='line', marker='x', linestyle='--', label='Previous Year (df_25, adjusted)')

plt.title('Total Amount by Week (Year-over-Year Comparison)')
plt.xlabel('Week (Aligned Year)')
plt.ylabel('Total Amount')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
